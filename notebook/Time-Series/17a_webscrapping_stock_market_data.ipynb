```python
import yfinance as yf

# Define ticker symbol
ticker_symbol = "AAPL"

# Get data on this ticker
ticker_data = yf.Ticker(ticker_symbol)

# Get the stock price history
tickerDf = ticker_data.history(
    start="2021-01-01",
    end="2021-12-31"
)

# Get the last closing price
lastPrice = tickerDf["Close"].iloc[-1]

# Print the last closing price
print(lastPrice)
```

```python
import yfinance as yf
import pandas as pd

# Define ticker symbol
tickerSymbol = "TSLA"

# Get data on this ticker
tickerData = yf.Ticker(tickerSymbol)

# Get the historical prices for this ticker
tickerDf = tickerData.history(
    start="2010-01-01",
    end="2020-01-26"
)

print(tickerDf.head())
print(tickerDf.shape)

if not tickerDf.empty:
    print(tickerDf["Close"].iloc[-1])
else:
    print("No historical data was returned.")
```

```python
import pandas as pd
import yfinance as yf
import datetime as dt
from datetime import date, timedelta

today = date.today()

d1 = today.strftime("%Y-%m-%d")
d1

d2 = (today - timedelta(days=365)).strftime("%Y-%m-%d")
d2

start_date = d2
end_date = d1

# Define ticker symbol
tickerSymbol = "META"

# Get data on this ticker
tickerData = yf.Ticker(tickerSymbol)

# Get the historical prices for this ticker
tickerDf = tickerData.history(
    start=start_date,
    end=end_date
)

tickerDf.head()

# Save DataFrame to CSV
# tickerDf.to_csv("../data/META.csv")
```
