```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Send a GET request to the Wikipedia page
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

headers = {
    "User-Agent": "WebScrapingLearning/1.0 (your-email@example.com)"
}

req = requests.get(url, headers=headers)

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(req.content, "html.parser")

# Find all wikitable tables
tables = soup.find_all("table", class_="wikitable")

print(f"Found {len(tables)} tables")

# Select the population table
table = tables[0]

# Find all the rows in the table
rows = table.find_all("tr")

# Create an empty DataFrame
data = pd.DataFrame(columns=["Rank", "Country", "Population"])

# Loop through each row and extract the data
for row in rows:

    # Find all the cells in the row
    cells = row.find_all("td")

    # Make sure the row has enough cells
    if len(cells) >= 3:

        # Extract the data from the cells
        rank = cells[0].text.strip()
        country = cells[1].text.strip()
        population = cells[2].text.strip()

        # Add the row to the DataFrame
        new_row = pd.DataFrame({
            "Rank": [rank],
            "Country": [country],
            "Population": [population]
        })

        data = pd.concat([data, new_row], ignore_index=True)

# Print the DataFrame
data
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Send a GET request to the Wikipedia page
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

headers = {
    "User-Agent": "WebScrapingLearning/1.0 (your-email@example.com)"
}

req = requests.get(url, headers=headers)

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(req.content, "html.parser")

# Find all wikitable tables
tables = soup.find_all("table", class_="wikitable")

print(f"Found {len(tables)} tables")

# Select the population table
table = tables[0]

# Find all the rows in the table
rows = table.find_all("tr")

# Create an empty DataFrame
data = pd.DataFrame(columns=["Rank", "Country", "Population"])

# Loop through each row and extract the data
for row in rows:

    # Find all the cells in the row
    cells = row.find_all("td")

    # Make sure the row has enough cells
    if len(cells) >= 3:

        # Extract rank
        rank = cells[0].text.strip()

        # Extract country name from <a> tag if available
        country_cell = cells[1].find("a")

        if country_cell is not None:
            country = country_cell.text.strip()
        else:
            country = cells[1].text.strip()

        # Extract population
        population = cells[2].text.strip()

        # Append the data to the DataFrame
        new_row = pd.DataFrame({
            "Rank": [rank],
            "Country": [country],
            "Population": [population]
        })

        data = pd.concat([data, new_row], ignore_index=True)

# Print the DataFrame
print(data)
```
