## Jena Climate Change Data Forecasting

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from prophet import Prophet
from prophet.plot import plot_plotly, plot_components_plotly
```

```python
# diata read karne ke liye
df = pd.read_csv('../data/jena_climate_2009_2016.csv')
df.head()
```

```python
df.shape
```

```python
# rename the columns to be used for prophet
df = df.rename(columns={'Date Time': 'ds', 'T (degC)': 'y'})
df.head()
```

```python
# Formate the datetime column
from pandas import to_datetime
df['ds'] = to_datetime(df['ds'], format='%d.%m.%Y %H:%M:%S')
```

```python
# Implement the Prophet model
model = Prophet()
model.fit(df)
forecast = model.make_future_dataframe(periods=365)
forecast = model.predict(forecast)
```

```python
# Plot that data
fig = model.plot_plotly(forecast)
fig.show()
```
